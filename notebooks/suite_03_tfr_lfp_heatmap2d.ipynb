# Suite 03: 2D Heatmaps of TFR Spectrograms
        
Generates 2D spectrogram heatmaps locked to stimulus onset across recorded areas.

Uses real preprocessed TFR arrays via `jnwb.session.OmissionSession.plot_tfr`
(backed by `tfr_from_preprocessed`) for session `sub-C31o_ses-230823`, areas
`V1` (probe C) and `FEF` (probe A), conditions `AAAB` (control) and `AAXB` (P2 omission).
`plot_tfr` returns `status='missing_tfr'` and no figure when a file is absent --
it does not fabricate random spectrograms.

```python
import os
import sys
import matplotlib.pyplot as plt
import numpy as np

sys.path.insert(0, os.path.abspath('..'))
import jnwb as oa
```

```python
nwb_path = 'D:/analysis/nwb/sub-C31o_ses-230823_rec.nwb'
assert os.path.exists(nwb_path), f'Missing NWB file: {nwb_path}'

session = oa.read(nwb_path)
print(session.summary())
```

```python
res_v1_control = session.plot_tfr(area='V1', condition='AAAB', phase=2)
print('V1 AAAB status:', res_v1_control['status'])
assert res_v1_control['status'] == 'completed', res_v1_control.get('error')
res_v1_control['figure']
```

```python
res_v1_omission = session.plot_tfr(area='V1', condition='AAXB', phase=3)
print('V1 AAXB status:', res_v1_omission['status'])
assert res_v1_omission['status'] == 'completed', res_v1_omission.get('error')
res_v1_omission['figure']
```

```python
res_fef_control = session.plot_tfr(area='FEF', condition='AAAB', phase=2)
print('FEF AAAB status:', res_fef_control['status'])
assert res_fef_control['status'] == 'completed', res_fef_control.get('error')
res_fef_control['figure']
```
