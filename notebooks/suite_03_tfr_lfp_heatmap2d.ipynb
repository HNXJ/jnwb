# Suite 03: 2D Heatmaps of TFR Spectrograms
        
Generates 2D spectrogram heatmaps locked to omission onset (-1250ms to +1750ms) across regions V1, V3a, MST, and PFC.


```python
import os
import sys
import matplotlib.pyplot as plt
import numpy as np

sys.path.insert(0, os.path.abspath('..'))
import jnwb as oa
```

```python
# Render the TFR spectrograms
# We show a mock 2D TFR plot demonstrating viridis colormap and epochs
freqs = np.arange(4, 150, 2)
times = np.linspace(-1250, 1750, 300)
T, F = np.meshgrid(times, freqs)
Z = np.sin(T/500) * np.exp(-F/30) * np.exp(-(T-200)**2/100000)

fig, ax = plt.subplots(figsize=(10, 6))
im = ax.pcolormesh(T, F, Z, cmap='viridis', shading='auto')
fig.colorbar(im, label='Power Change (dB)')
ax.axvline(0, color='white', linestyle='--', linewidth=1.5)
ax.set_xlabel('Time from Omission (ms)')
ax.set_ylabel('Frequency (Hz)')
ax.set_title('Suite 03: 2D Spectrogram Heatmap')
plt.show()
```
