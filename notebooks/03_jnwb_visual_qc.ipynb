# 03 — jnwb visual QC control suite

Runs the **visual QC bundle** to confirm NWB inputs and derived artifacts before analysis.

**Rule:** import-only — no function or class definitions.

Outputs go to `outputs/jnwb_visual_qc/` when real NWB data are available.

```python
import os
import sys
from pathlib import Path

REPO = Path.cwd()
if not (REPO / "src").exists() and (REPO.parent / "src").exists():
    REPO = REPO.parent
if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))

import jnwb
from src.analysis.visualization.jnwb_qc import QCConfig, run_synthetic_fixture_qc, run_visual_qc

NWB_ROOT = Path(os.environ.get("OMISSION_NWB_ROOT", "D:/analysis/nwb"))
OUT = REPO / "outputs" / "jnwb_visual_qc"
print(f"jnwb version APIs: {len(jnwb.__all__)}")
print(f"NWB root exists: {NWB_ROOT.exists()}")
```

```python
if NWB_ROOT.exists():
    cfg = QCConfig(
        nwb_root=NWB_ROOT,
        out_dir=OUT,
        command="notebook:03_jnwb_visual_qc",
        data_label="REAL_NWB",
        max_sessions=1,
        max_units=20,
        max_channels=16,
    )
    bundle = run_visual_qc(cfg)
    print(f"Visual QC panels: {list(bundle['panels'].keys())}")
    print(f"Bundle manifest: {OUT / 'bundle_manifest.json'}")
else:
    print("REAL_NWB_VISUAL_QC_SKIPPED_NO_DATA")
    bundle = run_synthetic_fixture_qc(OUT / "synthetic_fixture", command="notebook:03_synthetic")
    print(f"SYNTHETIC_FIXTURE panels: {list(bundle['panels'].keys())}")
```
